# UNIT 1 — Why We Study Principles of Programming Languages

## 1.1 What is a programming language?

A **programming language** is a formal language used to express instructions that can be processed by a computer.

Examples include Python, Java, C, C++, JavaScript, R, Ruby and Haskell.

Although their syntax differs, they all provide ways to express common ideas such as values, variables, expressions, decisions, repetition, functions, input and output.

````python
print("Hello, IIT1101!")
````

The underlying concept here is **output**. Other languages use different syntax:

```text
Python:  print("Hello")
Java:    System.out.println("Hello");
C:       printf("Hello");
```

PPL helps us see the concept behind the syntax.

## 1.2 Why study PPL?

PPL helps us:

1. understand how programming languages are structured;
2. compare languages;
3. learn new languages more easily;
4. choose suitable languages and paradigms;
5. understand how language design affects programs;
6. write clearer and more reliable software.

## 1.4 Variables

A **variable** is a name used to refer to a value.

```python
age = 18
```

`age` is the name and `18` is the value currently associated with it.

````python
age = 18
student_name = "Shalom"

print(age)
print(student_name)
````

Some languages require explicit declarations, for example:

```c

```

Python normally creates the name by assignment:

```python

```

The deeper PPL idea is **binding** — associating a name with a property or value.

## 1.5 Data types

A **data type** describes the kind of value and the operations that make sense for it.

- `int` — whole number
- `float` — decimal number
- `str` — text
- `bool` — `True` or `False`

# UNIT 2 — Programming Paradigms

## 2.1 What is a programming paradigm?

A **programming paradigm** is a general style or way of thinking about how a solution should be expressed.

A useful first distinction is:

### Imperative
Tell the computer **HOW** to perform the computation.

### Declarative
Describe **WHAT** result, rule or relationship is required.

A problem itself is not automatically imperative or declarative — the paradigm describes how we choose to express the solution.

## 2.2 Imperative programming — HOW

````python
marks = [45, 67, 81, 39, 55]
passed = []

for mark in marks:
    if mark >= 50:
        passed.append(mark)

print(passed)
````

This is imperative because we specify the steps and change program state. `passed` begins empty and is updated during execution.

## 2.3 Procedural programming

````python
def has_passed(mark):
    return mark >= 50

def show_result(name, mark):
    if has_passed(mark):
        print(name, "PASS")
    else:
        print(name, "FAIL")

show_result("Tariro", 67)
show_result("Rudo", 45)
````

Procedural programming organises a solution around **procedures/functions**.

```text
LARGE PROBLEM
     ├── procedure 1
     ├── procedure 2
     └── procedure 3
```

## 2.4 Object-oriented programming

````python
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    def has_passed(self):
        return self.mark >= 50

student1 = Student("Tariro", 67)

print(student1.name)
print(student1.has_passed())
````

OOP organises software around **objects** that combine state/data and behaviour/methods.

## 2.5 Parallel and concurrent programming

````python
from concurrent.futures import ThreadPoolExecutor
import time

def process_student(name):
    time.sleep(0.1)
    return f"Finished processing {name}"

students = ["A", "B", "C"]

with ThreadPoolExecutor(max_workers=3) as executor:
    results = list(executor.map(process_student, students))

for result in results:
    print(result)
````

**Concurrency** means tasks can make progress during the same period.  
**Parallelism** means computations can literally execute at the same time, usually on multiple cores/processors.

## 2.6 Declarative programming — WHAT

Instead of specifying every execution step, declarative programming focuses on the required result or relationship.

Three introductory branches are:

- Logic Programming
- Functional Programming
- Database / Query Processing

## 2.7 Logic programming

````python
# FACTS
marks = {
    "Tariro": 67,
    "Rudo": 45,
    "Anesu": 81
}

# RULE
def passes(student):
    return marks[student] >= 50

# QUERIES
print("Does Tariro pass?", passes("Tariro"))
print("Does Rudo pass?", passes("Rudo"))
````

Logic programming is commonly described as:

```text
FACTS + RULES + QUERIES
```

Python is not a logic-programming language, but the example makes the idea visible.

## 2.8 Functional programming

````python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Squares:", squares)
print("Even:", even_numbers)
````

Functional programming emphasises functions, expressions and data transformation. Python supports functional-style features but is a **multi-paradigm language**, not a purely functional language.

## 2.9 Database / query processing

````python
import sqlite3

connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE students (
    name TEXT,
    mark INTEGER
)
""")

cursor.executemany(
    "INSERT INTO students VALUES (?, ?)",
    [
        ("Tariro", 67),
        ("Rudo", 45),
        ("Anesu", 81),
        ("Tawanda", 39)
    ]
)

cursor.execute("""
SELECT name, mark
FROM students
WHERE mark >= 50
""")

print(cursor.fetchall())

connection.close()
````

The SQL query states **what rows are required**. The database engine decides how to retrieve them.

# UNIT 3 — Imperative Programming Paradigm

## 3.1 Sequence

````python
x = 10
y = 5
total = x + y
print(total)
````

Statements normally execute in order unless control flow changes that order.

## 3.2 Selection / decisions

````python
mark = 64

if mark >= 50:
    result = "PASS"
else:
    result = "FAIL"

print(result)
````

The Boolean expression `mark >= 50` produces `True` or `False`.

## 3.3 Repetition / iteration

````python
for number in range(1, 6):
    print(number)
````

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

`for` is useful when iterating through a collection or range.  
`while` is useful when repetition depends on a condition.

## 3.4 Procedures / functions

````python
def calculate_total(coursework, exam):
    return coursework + exam

print(calculate_total(35, 48))
print(calculate_total(28, 54))
````

A function provides reuse and abstraction by giving a meaningful name to a sequence of operations.

# OOP Concepts within the Imperative Paradigm

## 3.5 Encapsulation / data security

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount("Tariro", 100)
account.deposit(50)

print(account.owner)
print(account.get_balance())
````

Encapsulation groups data and behaviour and limits direct access to internal state. Python uses conventions and name-mangling rather than the same strict private-field model as Java.

## 3.6 Inheritance

````python
class Person:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Person: {self.name}"

class Student(Person):
    def __init__(self, name, programme):
        super().__init__(name)
        self.programme = programme

student = Student("Rudo", "Information Technology")

print(student.describe())
print(student.programme)
````

`Student` reuses and extends `Person`.

## 3.7 Abstraction

````python
def calculate_average(marks):
    return sum(marks) / len(marks)

marks = [60, 72, 81, 55]

print(calculate_average(marks))
````

The caller uses a meaningful operation without needing to know every internal step.

## 3.8 Polymorphism

````python
class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
````

The same method call `speak()` works with different object types.

# UNIT 4 — Declarative Programming and Language Quality

## 4.1 Declarative thinking

Declarative programming asks:

> What result or relationship should hold?

rather than:

> What exact sequence of state changes should be executed?

## 4.2 Functional transformation

````python
marks = [45, 67, 81, 39, 55]

passing_marks = list(filter(lambda mark: mark >= 50, marks))
adjusted_marks = list(map(lambda mark: mark + 5, marks))

print("Passing:", passing_marks)
print("Adjusted:", adjusted_marks)
````

## 4.3 Pure functions

````python
def square(x):
    return x * x

print(square(4))
print(square(4))
````

A **pure function** returns the same result for the same inputs and does not modify hidden external state.

````python
counter = 0

def next_value():
    global counter
    counter += 1
    return counter

print(next_value())
print(next_value())
````

This second function depends on and changes external state.

# Attributes of a Good Programming Language

## 4.4 Clarity

````python
# Less clear
x = 18
y = 52
z = x + y

# Clearer
coursework_mark = 18
exam_mark = 52
final_mark = coursework_mark + exam_mark

print(final_mark)
````

Clarity improves when code communicates its purpose.

## 4.5 Simplicity

````python
for number in [1, 2, 3]:
    print(number)
````

Simplicity means common ideas can be expressed without unnecessary complexity. It does not mean every program must be short.

## 4.6 Orthogonality

````python
print(len("IIT1101"))
print(len([10, 20, 30]))
print(len((1, 2, 3, 4)))
print(len({"a": 1, "b": 2}))
````

Orthogonality is about features combining in reasonably consistent ways with few special cases. The consistent use of `len()` across collection types is a simple illustration.

## 4.7 Naturalness

````python
mark = 63

if mark >= 50:
    result = "PASS"
else:
    result = "FAIL"

print(result)
````

The code expresses the decision in a way that closely resembles the problem statement.

## 4.8 Support for abstraction

````python
def final_result(coursework, exam):
    total = coursework + exam
    return "PASS" if total >= 50 else "FAIL"

print(final_result(25, 40))
````

Functions and classes hide implementation details behind meaningful interfaces.

## 4.9 Readability and writability

````python
numbers = [1, 2, 3, 4, 5]

# Compact
squares_1 = [x * x for x in numbers]

# More explicit
squares_2 = []
for x in numbers:
    squares_2.append(x * x)

print(squares_1)
print(squares_2)
````

**Readability** asks how easily code can be understood.  
**Writability** asks how easily a programmer can express a solution.

Very compact syntax may improve writability for experienced programmers while being less readable to beginners.

# UNIT 5 — Syntax, Semantics, Binding and Syntactic Criteria

## 5.1 Syntax

````python
valid_code = 'print("Hello")'
invalid_code = 'print("Hello"'

for source in [valid_code, invalid_code]:
    try:
        compile(source, "<student-example>", "exec")
        print(source, "-> syntactically valid")
    except SyntaxError:
        print(source, "-> SyntaxError")
````

A syntax error means code violates the grammatical rules of the language.

## 5.2 Semantics

````python
length = 10
width = 5

wrong_area = length + width
correct_area = length * width

print("Wrong meaning for area:", wrong_area)
print("Correct area:", correct_area)
````

Both expressions are valid Python. The first is **semantically/logically wrong for the intended problem**.

## 5.3 Syntax tree / Abstract Syntax Tree

````python
import ast

expression = "result = x * (y + z) - w"
tree = ast.parse(expression)

print(ast.dump(tree, indent=2))
````

A parser identifies structures such as assignment, multiplication, addition, subtraction and identifiers. Code is not treated as only a raw string of characters.

## 5.4 BNF-style grammar

A simplified grammar might be:

```text
<assignment> ::= <identifier> "=" <expression>
<expression> ::= <identifier>
               | <number>
               | <expression> "+" <expression>
               | <expression> "*" <expression>
```

For:

```python
total = coursework + exam
```

we can view:

```text
<assignment>
    ├── <identifier> → total
    ├── =
    └── <expression>
          ├── coursework
          ├── +
          └── exam
```

## 5.5 Symbol table

````python
course_code = "IIT1101"
pass_mark = 50

print("course_code" in globals())
print("pass_mark" in globals())
print("Value of course_code:", globals()["course_code"])
````

A **symbol table** associates identifiers with information such as value, type and scope. `globals()` is not Python's entire compiler symbol table, but it makes the name-to-information idea visible.

## 5.6 Binding

````python
x = 10
print("x =", x, "| type:", type(x))

x = "ten"
print("x =", x, "| type:", type(x))
````

A **binding** is an association between a program entity and a property, such as name ↔ value or name ↔ type.

## 5.7 Binding time

````python
value = 100
print(value, type(value))

value = 3.14
print(value, type(value))

value = "Data"
print(value, type(value))
````

**Binding time** describes when a binding is established. Possible times include language-design time, compile time, link time, load time and run time. Python performs many bindings at run time.

## 5.8 Static and dynamic typing

````python
value = 10
print(type(value))

value = "ten"
print(type(value))
````

In statically typed languages, many type checks happen before execution. In dynamically typed languages such as Python, many type relationships are determined and checked at run time.

## 5.9 Implicit language operations

````python
class Course:
    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)

course = Course(["A", "B", "C"])

print(len(course))
````

`len(course)` causes Python to use the object's `__len__()` method behind the scenes. Surface syntax can trigger deeper runtime operations.

## 5.10 Error checking

````python
try:
    result = 10 / 0
except ZeroDivisionError as error:
    print("Runtime error caught:", error)
````

````python
length = 8
width = 4

area = length + width   # valid syntax, wrong formula
print("Incorrect area:", area)
````

Useful distinction:

1. **Syntax error** — grammar is invalid.
2. **Runtime error** — execution begins but an invalid operation occurs.
3. **Semantic/logical error** — program runs but implements the wrong meaning.

## 5.11 Verification

````python
def add(a, b):
    return a + b

assert add(2, 3) == 5
assert add(-1, 1) == 0

print("Verification checks passed.")
````

Verification asks:

> **Are we building the program correctly according to its specification?**

## 5.12 Validation

````python
def simple_average(coursework, exam):
    return (coursework + exam) / 2

def weighted_final(coursework, exam):
    return coursework * 0.4 + exam * 0.6

coursework = 70
exam = 50

print("Simple average:", simple_average(coursework, exam))
print("Required weighted mark:", weighted_final(coursework, exam))
````

Validation asks:

> **Are we building the correct program for the real user's need?**

A program can be technically correct but still solve the wrong problem.

## 5.13 Readability, writability and verifiability

````python
def is_pass(mark):
    """Return True when mark satisfies the pass rule."""
    return mark >= 50

test_marks = [49, 50, 75]

for mark in test_marks:
    print(mark, is_pass(mark))

assert is_pass(49) is False
assert is_pass(50) is True
assert is_pass(75) is True
````

- **Readability** — easy to understand.
- **Writability** — easy to express a solution.
- **Verifiability** — practical to reason about and test correctness.
